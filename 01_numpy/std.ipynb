```python
# 표준편차는 값들이 평균 주변에 얼마나 흩어져 있는지 나타내는 숫자
import numpy as np

# 둘 다 평균은 0.7이지만, a는 값이 전부 같고 b는 퍼져있음.
a = np.array([0.7, 0.7, 0.7])
b = np.array([0.4, 0.7, 1.0])

# 0.0( 1.1102230246251565e-16 )
print(a.std())
# 약 0.245
print(b.std())

# 값들이 모여 있으면 표준편차가 작고, 퍼져 있으면 커져
# 대화 품질 점수라면 평균이 같아도 응답마다 점수 차이가 얼마나 나는지 볼 수 있다.

# 그럼 아래 둘 중 표준편차가 더 큰 족은 어느쪽일까?
c = np.array([0.6, 0.7, 0.8])
d = np.array([0.1, 0.7, 1.3])

#정답은 d

#둘 다 평균은 0.7이지만 d의 값들이 평균에서 더 멀리 떨어져 있으니 표준편차가 더 크다.
print(c.std())  # 약 0.082
print(d.std())  # 약 0.490
```

```python
# Numpy를 호출
import numpy as np

# 1차원 배열 생성
scores = np.array([0.91, 0.32, 0.82, 0.95, 0.42, 0.83, 0.68])

print(scores.mean())
print(scores.std())
```

```python
# 연습문제 1
import numpy as np

# 각 행은 대화 하나, 각 열은 자연스러움·구체성·안전성
scores = np.array([
    [0.9, 0.8, 1.0],
    [0.4, 0.7, 0.9],
    [0.8, 0.9, 0.8]
])
# 각 대화(행)의 점수 평균을 구해서 새 배열에 저장
conversation_mean = scores.mean(axis=1)
# 각 평균이 0.8 이상인지 확인해서 True/False 배열 생성
is_passed = conversation_mean >= 0.8
# True/False 배열을 보고, True인 자리의 행을 scores에서 가져옴
passed_conversations = scores[is_passed]

# 출력
print(passed_conversations)
```

```python
# 연습문제 2
import numpy as np

# 각 행은 대화 하나, 각 열은 자연스러움·구체성·안전성
scores = np.array([
    [0.9, 0.8, 1.0],
    [0.4, 0.7, 0.9],
    [0.8, 0.9, 0.8]
])

naturalness = scores[:, 0]
# [0.9, 0.4, 0.8]

is_natural = naturalness >= 0.8
natural_conversations = scores[is_natural]

print(natural_conversations)
```
