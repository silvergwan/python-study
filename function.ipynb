```python
# 함수란 자주 쓰는 코드를 이름 붙여서 묶어두는 것

# 예를 들어 이런 코드가 있다면
print("안녕하세요 Kim님")
print("안녕하세요 Lee님")
print("안녕하세요 Park님")
print("=" * 15)

# 사람이 늘어나면 코드도 늘어남

# 그래서 함수로 묶음
def greet(name):
    print(f"안녕하세요 {name}님")
    
# 그리고 필요할 때 호출
greet("Kim")
greet("Lee")
greet("Park")
greet("Choi")
```

```python
# def가 뭐냐

# 쪼개보면

# def
# → "나 함수 만들 거야"

# greet
# → 함수 이름

# (name)
# → 함수가 받을 데이터

# :
# → 이제 함수 내용 시작

# 그리고 들여쓰기 안쪽이 함수가 실행할 코드

def greet(name):
    print(f"안녕하세요 {name}님")

# greet이라는 이름을 받고 "안녕하세요 {이름}님을 출력하는 함수"
```

```python
# 연습문제 1

# 숫자를 매개변수로 받는 홀 짝 구분하는 함수 선언
def check_odd_even(number):
    # 만약 숫자 나누기 2의 나머지 값이 0이면(짝수면)
    if number % 2 == 0:
        # 짝수 리턴
        return "짝수"
    # 아니면
    else:
        # 홀수 리턴
        return "홀수"
    
print(check_odd_even(10))
print(check_odd_even(7))
```

```python
# 연습문제 2

students = [
    {"name": "Kim", "score": 80},
    {"name": "Lee", "score": 55},
    {"name": "Park", "score": 95}
]

# 학생 딕셔너리를 매개변수로 받는 합격 학생 수 확인 함수
def count_pass_students(students):
    # 지역 변수 선언
    cnt = 0
    # students 딕셔너리 student로 순회
    for student in students:
        # 만약 student의 점수가 60이상이면
        if student["score"] >= 60:
            # 합격 학생 수 + 1
            cnt += 1
    # 순회 다 하고 합격 학생 수 리턴
    return cnt

print(count_pass_students(students))
```

```python
# 연습 문제
# 학생 리스트를 받아서 합격한 학생들의 이름만 담긴 리스트를 return하는 함수

students = [
    {"name": "Kim", "score": 80},
    {"name": "Lee", "score": 55},
    {"name": "Park", "score": 95}
]
# students 딕셔너리를 매겨변수로 받는 합격 학생 이름 반환 함수
def get_pass_students(students):
    # 합격한 학생이 들어갈 빈 배열 선언
    passed_student = []
    # students 딕셔너리 student로 순회
    for student in students:
        # 만약 student의 점수가 60점 이상이면
        if student["score"] >= 60:
            # 빈 배열에 합격학생의 이름 append
            passed_student.append(student["name"])
    # 순회 다 하고 이름이 들어있는 배열 리턴
    return passed_student

print(get_pass_students(students))
```
