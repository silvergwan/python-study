```python
# enumerate가 뭐냐

# 리스트가 있다고 치자
names = ["Kim", "Lee", "Park"]

# 그냥 반복하면
for name in names:
    print(name)
    
# 결과
# Kim
# Lee
# Park
```

```python
#근데 몇 번째 데이터인지도 같이 필요한 경우가 분명히 있다
# 그럴 때는 enumrate를 쓴다

names = ["Kim", "Lee", "Park"]

for index, name in enumerate(names):
    print(index, name)
    
# 이런 식으로 번호 + 값을 같이 준다.

print("="*10)

#번호를 1부터 시작하고 싶다면
for index, name in enumerate(names, start=1):
    print(index, name)
```

```python
students = [
    {"name": "Kim", "score": 80},
    {"name": "Lee", "score": 55},
    {"name": "Park", "score": 95}
]

for index, student in enumerate(students, start=1):
    print(index, student["name"], student["score"])
```
