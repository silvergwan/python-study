```python
# 연습문제 3-1(불린 인덱싱)
# numpy 호출
import numpy as np

# 1차원 배열 생성
scores = np.array([50, 70, 80, 90])

# 불린 인덱싱을 이용해 70점 이상인 점수에 True, 나머지 점수에 False를 저장한 배열을 만듦
ispassed = scores >= 70

# 불린 인덱싱을 사용해 70점 이상인 점수, 즉 True에 해당하는 필드만 저장하는 새 배열을 만듦
passed_scores = scores[ispassed]
print(passed_scores)
```

```python
# 연습문제 3-2(불린 인덱싱 2)
# numpy 호출
import numpy as np

# 1차원 배열 생성
scores = np.array([50, 70, 80, 90])

# 불린 인덱싱을 이용해 70점 이상인 점수에 True, 나머지는 False를 저장한 새 배열을 보고 True에 해당하는 필드만 저장한 새 배열을 만듦
scores = scores[scores >= 70]
# 출력
print(scores)
```
