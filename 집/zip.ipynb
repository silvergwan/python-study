```python
# zip은 여러 리스트를 같은 위치끼리 묶어서 반복할 때 쓴다.

names = ["Kim", "Lee", "Park"]
scores = [80, 55, 95]

for name, score in zip(names, scores):
    print(name, score)
    
# zip(names, scores)를 머릿속으로 보면
# ("Kim", 80)
# ("Lee", 55)
# ("Park", 95)
# 이렇게 묶임
```

```python
# 머신러닝에서도 이게 꽤 중요하다고 하는데

#예를 들어

texts = [
    "이 영화 재밌다",
    "별로였다",
    "또 보고 싶다"
]

labels = [1, 0, 1]

# 입력 데이터와 정답이 따로 존재 가능

for texts, labels in zip(texts, labels):
    print(texts, labels)
```

```python
# 연습문제 1

names = ["Kim", "Lee", "Park"]
scores = [80, 55, 95]

for name, score in zip(names, scores):
    if score >= 60:
        print(name, score)
```
