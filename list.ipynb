```python
# 리스트를 사용하면 1, 3, 5, 7, 9의 숫자 모음을 다음과 같이 간단하게 표현할 수 있다.
odd = [1, 3, 5, 7, 9]

# 리스트를 만들 때는 위에서 보는 것과 같이 대괄호([])로 감싸 주고 각 요솟값은 쉼표(,)로 구분해 준다.

리스트명 = [요소1, 요소2, 요소3, ...]
```

```python
# 리스트의 인덱싱
# 리스트 역시 문자열처럼 인덱싱을 적용할 수 있다. 먼저 a 변수에 [1, 2, 3] 값을 설정한다.

a = [1, 2, 3]
print(a)

# a[0]은 리스트 a의 첫 번째 요솟값(1)을 말한다.
print(a[0])

# 다음 예는 리스트의 첫 번째 요소인 a[0]과 세 번째 욧인 a[2]의 값을 더한 것이다.
print(a[0]+a[2])

# a[-1]은 리스트 a의 마지막 요솟값을 말한다.
print(a[-1])
```

```python
# 리스트의 슬라이싱
a = [1, 2, 3, 4, 5]
print(a[0:2])
```

```python
# 리스트 연산하기

# 리스트 더하기
a = [1, 2, 3]
b = [4, 5, 6]
result = a + b
print(result)

# 리스트 반복하기(곱하기)
a_star = a * 3
print(a_star)
```

```python
# 리스트 길이 구하기

a = [1, 2, 3]

print(len(a))
```
