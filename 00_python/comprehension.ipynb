```python
# 리스트 컴프리헨션이란?
# for문으로 새 리스트 만드는 걸 짧게 쓰는 문법

students = [
    {"name": "Kim", "score": 80},
    {"name": "Lee", "score": 55},
    {"name": "Park", "score": 95}
]

# 아까 내가 쓴 코드
passed_student = []

for student in students:
    if student["score"] >= 60:
        passed_student.append(student["name"])
        
# 리스트 컴프리헨션으로 바꾸면?        
passed_student = [
    student["name"]
    for student in students
    if student["score"] >= 60
]

# 한 줄로 쓰면     # 3번           # 1번                   # 2번
passed_student = [student["name"] for student in students if student["score"] >= 60]


# students에서 student를 하나씩 꺼내서
# → score가 60 이상이면
# → student["name"]을 새 리스트에 넣어라
```

```python
# 1, 2, 3, 4가 들어있는 배열 선언
numbers = [1, 2, 3, 4]

# result는 numbers에서 number를 하나씩 꺼내서
# number * 2 시켜주고 result 리스트에 넣음
result = [number * 2 for number in numbers]

print(result)
```

```python
numbers = [1, 2, 3, 4, 5, 6]

# result는 numbers에서 number를 하나씩 꺼내 조건문(짝수찾기)를 거쳐 통과한 숫자만 result에 들어간다.
result = [number for number in numbers if number % 2 == 0]

print(result)
```

```python
numbers = [1, 2, 3, 4, 5]
# result는 numbers에서 number를 하나씩 꺼내 조건문(짝수찾기)를 거쳐 통과한 숫자만 * 10을 해서 result에 들어간다.
result = [number * 10 for number in numbers if number % 2 ==0]

print(result)
```

```python
numbers = [3, 10, 15, 20, 25, 30]
# result는 numbers에서 number를 하나씩 꺼내어 조건문(10이상이고 짝수인) number를 * 2시켜서 result 배열에 넣는다.
result = [number * 2 for number in numbers if number >= 10 and number % 2 == 0]

print(result)
```
