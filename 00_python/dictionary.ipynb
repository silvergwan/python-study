```python
# 딕셔너리

# 이름표가 붙은 서랍장

#리스트는 숫자로 값을 찾지만, 딕셔너리는 이름으로 매칭되는 값을 찾는다.

student = {
    "name": "은관",
    "score": 90,
    "age": 18
}

print(student["name"])

#결과
# 은관

# 즉 구조는
# key        value

# "name"  →  "은관"
# "score" →  90
# "age"   →  18
```

```python
# 값 가져오기

student = {
    "name": "은관",
    "score": 90
}

print(student["name"])
print(student["score"])

# 결과

# 은관
# 90
```

```python
# 딕셔너리 + 반복문

student = {
    "name": "은관",
    "score": 90,
    "age": 18
}

for key in student:
    print(key)

# 키 값 나옴    
# name 
# score 
# age

#값까지 가져오고 싶으면?
for key, value in student.items():
    print(key, value)
    
# 값만 가져오고 싶으면?

for value in student.values():
    print(value)
```

```python
# 연습 문제 1
# 나이가 20살 이상인 사람의 이름만 출력하는 코드

users = [
    {"name": "Kim", "age": 20},
    {"name": "Lee", "age": 17},
    {"name": "Park", "age": 23}
]

for user in users:
    if(user["age"] >= 20):
        print(user["name"])
        
        
```

```python
# 연습 문제 2
# 점수가 60점 이상인 학생의 이름과 점수를 같이 출력

students = [
    {"name": "Kim", "score": 80},
    {"name": "Lee", "score": 55},
    {"name": "Park", "score": 95}
]

for student in students:
    if(student["score"] >= 60):
        print(f"{student["name"]} {student["score"]}")
```

```python
# 연습 문제 3
# 60점 이상인 학생들의 점수 평균을 출력

students = [
    {"name": "Kim", "score": 80},
    {"name": "Lee", "score": 55},
    {"name": "Park", "score": 95}
]

count = 0
sum = 0

for student in students:
    if(student["score"] >= 60):
        count += 1
        sum += student["score"]

avg = sum / count

print(avg)
```

```python
# 연습문제 4
# 가장 점수가 높은 학생의 이름과 점수를 출력

students = [
    {"name": "Kim", "score": 80},
    {"name": "Lee", "score": 55},
    {"name": "Park", "score": 95}
]

max_score = 0
max_name = ""

for student in students:
    if(student["score"] >= max_score):
        max_score = student["score"]
        max_name = student["name"]

print(max_score)
```
