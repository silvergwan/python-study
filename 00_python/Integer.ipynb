```python
# 정수형(integer)은 말 그대로 정수를 뜻하는 자료형이다. 다음은 양의 정수, 음의 정수, 숫자 0을 변수 a에 대입하는 예이다.

a = 123
a = -178
a = 0
```

```python
# 실수형(floating-point)은 소수점이 포함된 숫자를 말한다. 다음은 실수를 변수 a에 대입하는 예이다.

a = 1.2
a = -3.45

print(a)
```

```python
# 사칙 연산

a = 3
b = 4
a + b
# 7

a - b
# -1

a * b
# 12

a / b
# 0.75

```

```python
# x의 y제곱을 나타내는 ** 연산자
# x ** y처럼 사용하면 x의 y제곱 값을 반환한다.

a = 3
b = 4
a ** b
# 81
```

```python
# 나눗셈 후 나머지를 반환하는 % 연산자

7 % 3
# 1

3 % 7
# 3
```

```python
# 나눗셈 후 몫을 반환하는 // 연산자
# / 연산자를 사용하여 7 나누기 4를 하면 결과는 1.75가 된다. 반면 // 연산자를 사용하면 나눗셈 후 몫만 반환한다. 두 연산자의 차이를 비교해 보자.

7 / 4
# 1.75

7 // 4
# 1
```

```python
# 문제 1. 몫과 나머지 구하기
# 다음 코드를 실행했을 때 출력되는 a와 b의 값을 각각 구하시오.

a = 17 // 5
b = 17 % 5

# a = 3, b = 2
print(a, b)
```

```python
# 문제 2. 연산자 우선순위와 거듭제곱
# 다음 코드의 실행 결과는 얼마일까요?

x = 2
y = 3

result = x ** y + 10 / 2

# result = 13
# 오답: 8을 5.0과 더했기 때문에 13.0이 나온다.
print(result)
```

```python
# 문제 3. 복합 연산자 이해하기
# 다음 코드가 실행된 후 최종 num 변수에 저장된 값은 얼마일까요?

num = 10
num += 5
num *= 2
num -= 4

# num = 26
print(num)
```
